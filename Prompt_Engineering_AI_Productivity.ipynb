## 1. Zero-shot Prompting

Zero-shot prompting means asking an AI model to perform a task without giving examples.

**Example:** Asking AI to explain Python variables without showing sample answers.

```python
prompt = "Explain Python variables in simple English."

print("Prompt:", prompt)
```

**Observation:**

I learned that zero-shot prompting allows AI to perform a task using only the instructions provided, without examples.

## 2. Few-shot Prompting

Few-shot prompting means providing a few examples to AI so it can understand the expected output format.

**Example:** Giving AI examples of positive and negative customer reviews before asking it to classify a new review.

```python
prompt = """
Classify the following reviews as Positive or Negative.

Example 1:
Review: I love this product.
Answer: Positive

Example 2:
Review: The product is very bad.
Answer: Negative

Now classify:
Review: The quality is excellent.
"""

print(prompt)
```

**Observation:**

The AI uses the given examples to understand the classification pattern and classify a new review as Positive or Negative.

## 3. Role-based Prompting

Role-based prompting means giving AI a specific role to get more relevant responses.

**Example:** Asking AI to act as a Python teacher and explain Python concepts in simple English.

```python
prompt = """
Act as a Python teacher.
Explain Python variables in simple English.
Give 3 examples with outputs.
"""

print(prompt)
```

**Observation:**

I learned that role-based prompting helps AI respond according to the role and instructions provided.

## 4. Context and Constraints

Context provides background information about a task.

Constraints specify rules or limitations that AI should follow.

**Example:** Asking AI to explain Python to a beginner in simple English using only 5 sentences.

```python
prompt = """
Context: I am a beginner learning Python.

Instruction: Explain Python loops.

Constraints:
1. Use simple English.
2. Give only 2 examples.
3. Keep the explanation short.
"""

print(prompt)
```

**Observation:**

I learned that context provides background information, while constraints control how AI generates its response.

## 5. Structured Output using JSON

JSON stands for JavaScript Object Notation.

It is used to organize and exchange data in a structured format.

**Example:** Student details stored using JSON.

```python
import json

student = {
    "name": "Indhu",
    "course": "BTech ECE",
    "skills": ["Python", "SQL", "Machine Learning"],
    "year": 2025
}

json_output = json.dumps(student, indent=4)

print(json_output)
```

**Observation:**

I learned how to create structured JSON output using Python. JSON organizes information in key-value pairs.

## 6. Text Summarization

Text summarization is the process of reducing long text into a shorter version while keeping the important information.

```python
text = """
Artificial Intelligence is a technology that enables
machines to perform tasks that normally require human intelligence.
It is used in healthcare, education, banking and many other fields.
AI helps save time, improve productivity and solve problems.
"""

summary = """
Artificial Intelligence helps machines perform intelligent tasks.
It is used in many industries to save time and solve problems.
"""

print("Original Text:")
print(text)

print("\nSummary:")
print(summary)
```

**Observation:**

I learned how summarization converts long information into short and meaningful content while keeping the main points.

## 7. Text Classification

Text classification is the process of categorizing text into predefined labels based on its meaning.

```python
reviews = [
    "This product is excellent",
    "This product is very bad",
    "I love this product"
]

for review in reviews:
    if "excellent" in review or "love" in review:
        category = "Positive"
    else:
        category = "Negative"

    print("Review:", review)
    print("Category:", category)
    print()
```

**Observation:**

I learned how to classify text into different categories such as Positive and Negative using Python.

## 8. AI-assisted Coding and Debugging

AI-assisted coding means using AI tools to generate, explain and improve code.

Debugging means identifying and correcting errors in a program.

**Example:** Asking AI to find and correct errors in Python code.

```python
# Python program with an error

a = 10
b = 5

result = a + b

print("The result is:", result)
```

```python
# Correct the error

a = 10
b = 5

result = a + b

print("The result is:", result)
```

**Observation:**

I learned how AI can assist programmers in writing, explaining and debugging Python code.

## 9. Prompt Evaluation and Testing

Prompt evaluation means checking whether a prompt generates accurate, relevant and useful responses.

Prompt testing helps us compare different prompts and improve their quality.

```python
prompts = [
    "Explain Python.",
    "Explain Python variables in simple English with 2 examples."
]

for i in range(len(prompts)):
    print("Prompt", i + 1, ":")
    print(prompts[i])
    print()
```

**Prompt Testing Report**

| Prompt | Clarity | Expected usefulness |
|---|---|---|
| Explain Python | Low | General explanation |
| Explain Python variables in simple English with 2 examples | High | Focused explanation |

**Observation:**

I learned that clear and specific prompts produce more focused and useful responses. This example compares prompt quality, not actual AI response accuracy.

## 10. AI Productivity Toolkit

An AI Productivity Toolkit is a collection of reusable prompts that helps people complete daily tasks efficiently.

It includes:

1. Summarization Assistant
2. Email Writing Assistant
3. Resume Assistant
4. Coding Assistant
5. Research Assistant

```python
toolkit = {
    "Summarization Assistant":
        "Summarize the given text in 5 simple points.",

    "Email Writing Assistant":
        "Write a professional email in simple English.",

    "Resume Assistant":
        "Create a professional fresher resume.",

    "Coding Assistant":
        "Explain and debug the given Python code.",

    "Research Assistant":
        "Explain the given topic with important points."
}

for tool, prompt in toolkit.items():
    print("\n", tool)
    print("Prompt:", prompt)
```

**Observation:**

I created an AI Productivity Toolkit containing reusable prompts for summarization, email writing, resume building, coding and research assistance.
